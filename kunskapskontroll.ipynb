1. Vad är skillnaden på en *tuple* och *list* i Python? Är den ena datatypen bättre än den andra? 

>**Svar:** <br>
En tuple är immutable och får inte ändra på sin data efter att den skapats. <br><br>
En list är immutable och fär ändra i sin data hur mycket den vill. <br><br>
Ingen datatyp är bättre än den andra, vilken man bör välja beror på vad man vill att den ska kunna göra, eller inte. Är det viktigt att innehållet inte ändras väljer man en tuple. Behöver man kunna ändra, lägga till eller ta bort data är en list ett bättre val.

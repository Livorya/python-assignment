1. Vad är skillnaden på en *tuple* och *list* i Python? Är den ena datatypen bättre än den andra? 

>**Svar:** <br>
En tuple är immutable och får inte ändra på sin data efter att den skapats. 
<br><br>
En list är immutable och fär ändra i sin data hur mycket den vill. 
<br><br>
Ingen datatyp är bättre än den andra, vilken man bör välja beror på vad man vill att den ska kunna (eller **inte** kunna) göra. Är det viktigt att innehållet inte ändras väljer man en tuple. Behöver man kunna ändra, lägga till eller ta bort data är en list ett bättre val.

5. I Python ska du nu: <br>
  a) Skapa en klass som heter BankAccount. Klassen skall ha attributet "account_holder" som visar kontoinnehavarens namn samt attributet "balance" som visar kontoinnehavarens balans. Klassen skall ha metoden "deposit()" för att kunna sätta in pengar på kontot samt metoden "withdraw()" för att kunna ta ut pengar från kontot. Om bankinnehavaren försöker ta ut mer pengar än vad som finns på kontot skall meddelandet "Too low balance" printas ut. 
  <br><br>
  b) Skapa en instans av klassen och testa så klassen funkar så som du förväntar dig. Du kan till exempel prova printa ut attributen, sätta in pengar och ta ut pengar.

```python
# a)
class BankAccount:
  '''Creates a bank account object that needs a name of the account holder and have a starting balance of 0$'''
  def __init__(self, account_holder):
    self.account_holder = account_holder
    self.balance = 0
  
  def deposit(self, sum):
    '''Deposits the sum given into the balance of the account holder.'''
    self.balance += sum
    print(f"{sum}$ has been added to {self.account_holder}'s account, new balance {self.balance}$")
  
  def withdraw(self, sum):
    '''If possible withdraws the sum from the balance of the account holder, otherwise prints a "To low balance" message'''
    if self.balance >= sum:
      self.balance -= sum
      print(f"{sum}$ has been withdrawn from {self.account_holder}'s account, new balance {self.balance}$")
    else:
      print(f"Cannot withdraw {sum}$, to low balance in {self.account_holder}'s account, current balance {self.balance}$")

  def viewBalance(self):
    '''Prints a message of the balance and returns the value.'''
    print(f"{self.account_holder} got a {self.balance}$ balance in his account")
    return self.balance
```

```python
# b)
my_account = BankAccount("Bob")

my_account.deposit(1000)
my_account.viewBalance()
my_account.withdraw(600)
my_account.withdraw(500)
```

6. På intervjuer brukar ibland kodtest genomföras. Här är ett exempel på en fråga som hade kunnat dyka upp på ett kodtest. <br>
Skapa en Python funktion som beräknar antalet vokaler i en sträng. 

    Vokalerna är: **A**, **E**, **I**, **O**, **U**, **Y**, **Å**, **Ä** och **Ö**. 

    Exempelvis: vowel_checker(’hjärna’) ska returnera ”2” eftersom ordet ”hjärna” har de två vokalerna ”ä” och ”a”. 

```python
def vowel_checker(word):
  '''Counts and returns the number of vowels in a word.'''
  vowels = ("A", "E", "I", "O", "U", "Y", "Å", "Ä", "Ö")
  count = 0
  for letter in word:
    if letter.upper() in vowels:
      count += 1
  return count
```

```python
word = "hjärna"
print(f'"{word}" has {vowel_checker(word)} vowels')

word = "HelloWorld!"
print(f'"{word}" has {vowel_checker(word)} vowels')

word = "Floccinaucinihilipilification"
print(f'"{word}" has {vowel_checker(word)} vowels')
```

7. På intervjuer brukar ibland kodtest genomföras. Här är ett exempel på en fråga som hade kunnat dyka upp på ett kodtest. <br>
Skapa en funktion som tar två listor som argument och returnerar de gemensamma elementen i listorna i en ny lista.  

    Exempelvis:  
    `common_elements([4, ’apple’, 10, ’hi’, 3], ([23 ’apple’, 5, 9, 3])` ska returnera svaret `[’apple’, 3]`.

```python
def common_elements(list1, list2):
  '''Takes two lists and return a new one with the common elements.'''
  if isinstance(list1, list) and isinstance(list2, list):
    return list(set(list1).intersection(set(list2)))
  else:
    print('Both arguments need to be of type "list"')
```

```python
common_elements([4, "apple", 10, "hi", 3], [23, "apple", 5, 9, 3])
```

##### Lärarens exempel (att insiperars av)

```python
df = pd.read_csv("data/diamonds.csv", index_col=False)

df["cut"] = pd.Categorical(df.cut, ["Fair", "Good", "Very Good", "Premium", "Ideal"], ordered=True)
df["clarity"] = pd.Categorical(df.clarity, ["I1", "SI2", "SI1", "VS2", "VS1", "VVS2", "VVS1", "IF"], ordered=True)
df["color"] = pd.Categorical(df.color, " ".join("JIHGFED").split(), ordered=True)

df = df[(df.y < 30) & (df.z < 30) & (df.x > 0) & (df.y > 0) & (df.z > 0)].copy()
```

```python
df["color_encoded"] = df.color.cat.codes
df["clarity_encoded"] = df.clarity.cat.codes
df["cut_encoded"] = df.cut.cat.codes
```

```python
df.color
```

```python
df[["clarity_encoded", "price"]].corr()
```

```python
df[["cut_encoded", "price"]].corr()
```

```python
df["carat_bins"] = pd.cut(df.carat, bins=10)

carat_groups = df.groupby("carat_bins", observed=False)

carat_groups.groups
```

```python
corrs = []

for i, group in carat_groups:
  corrs.append((i, len(group), group.corr(numeric_only=True)))

corrs[0][2]["price"]
```

```python
fig, axs = plt.subplots(5, 2, figsize=(12, 18))

for data, ax in zip(corrs, axs.flatten()):
  sns.heatmap(data[2].price.to_frame(), annot=True, ax=ax)
  ax.set(title=f"{data[0]} {data[1]}")
```
